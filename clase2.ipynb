## Estructuras de repetición:
Es una estructura computacional que permite repetir una acción n cantidad de veces hasta que se logre una condición

# Estructura PARA (for)
Estructurs rmpleada cuando se sabe cuantas veces se debe repetir una acción.

````python
for intento in range (31):
  if intento%3==0:
    print(intento)
````

````python
n=int(input("Ingrese un número: "))
for intento in range(11):
  res = intento*n
  print(f"{n} x {intento} = {res}")
````

````python
for i in range(11):
  print(f"tabla del {i}")
  for m in range(11):
    print(f"{i} x {m} = {m*i}")
````

````python
n=int(input("Ingrese un número: "))
for i in range(1,n+1):
  if i==1:
    print("1", end="  ")
  else:
    if i%2==0:
     print(f"-1/{i}",end=" ")
    else:
      print(f"+1/{i}",end=" ")
````

## ESTRUCTURA MIENTRAS (WHILE)
Estructura computacional de repetición, empleada para repetir una acción **mientras** se cumple una condicion.



```
while (condición):
  Ejecuta el código
  Modificador de la condición
  break()
```



````python
num =2
while num <=20:
  print(num)
  num=num+2
````

````python
i=1
n=1
while n <=10:
  while i<=10:
   print(f"{n} x {i} ={i*n}")
   i=i+1
  n=n+1
  i=1
````
