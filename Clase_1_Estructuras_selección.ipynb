##If/elif/else
Evalúa tantas condiciones como elif hayan. Si no se cumple, pasa a la siguiente condicion hasta llegar al else

Sintaxis

```
if(condicion1):
 aquí se ejecuta el código si condición es V
elif(condición2):
 se ejecuta si condición1 es falsa pero condición2 es V
elif (condición3):
 se ejecuta si condición 1 y 2 son falsas pero condición3 es V
else:
 se ejecuta si ninguna condición es V
```



##Match/case
La estructura match evalúa una variable frente a varios bloques case. El comodín case __ funciona como ninguna de las anteriores.

Sintaxis

```
match(condición):
 case v1: aqui se ejecuta condición==v1
 case v2: aquí se ejecuta condición==v2
 case v3: aquí se ejecuta condición==v3
 case__:ninguna de las anteriores
```
